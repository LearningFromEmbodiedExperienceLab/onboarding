```python
import robotics
from robotics.ik.differential import damped_lstsq

import numpy as np
import matplotlib.pyplot as plt

print("python:", __import__("sys").executable)
print("robotics:", robotics.__file__)
```

```python
J = np.array(
    [
        [1.0, 1.0, 1.0],
        [2.0, 2.0, 2.0],
        [3.0, 3.0, 3.0],
    ]
)
dx = np.array([1.0, 0.0, 0.0])

dampings = np.logspace(-6, 1, 40)
norms = []
dq_norms = []
for lam in dampings:
    dq = damped_lstsq(J, dx, damping=float(lam))
    norms.append(np.linalg.norm(J @ dq - dx))
    dq_norms.append(np.linalg.norm(dq))

fig, ax = plt.subplots()
ax.loglog(dampings, norms, label=r"$\|J\\,dq - dx\|$")
ax.loglog(dampings, dq_norms, label=r"$\|dq\|$")
ax.set_xlabel("damping $\\lambda$")
ax.legend()
ax.set_title("Rank-1 J: more damping → smaller, more stable step")
fig
```
